## Deployment Target

The trained Random Forest model will be deployed using Amazon SageMaker Endpoint.  
The endpoint will accept time-based ride demand features as input and return predicted hourly ride demand.

Input:
- pickup_hour
- pickup_day
- is_weekend
- is_morning_peak
- is_evening_peak

Output:
- predicted hourly ride demand

```python
def predict_ride_demand(pickup_hour, pickup_day, is_weekend, is_morning_peak, is_evening_peak):
    input_data = pd.DataFrame([{
        "pickup_hour": pickup_hour,
        "pickup_day": pickup_day,
        "is_weekend": is_weekend,
        "is_morning_peak": is_morning_peak,
        "is_evening_peak": is_evening_peak
    }])

    prediction = model.predict(input_data)
    return int(prediction[0])
```
